### Module 06 – Single-Target Tracking Demonstration on a MOT16 Test Sequence

**Objective.** This notebook produces the course's required deliverable: a tracking video of at least one target pedestrian on a MOT16 test sequence, using the best configuration found in Notebooks 04 and 05.

**Procedure**
1. Select a test sequence (for example MOT16-01, 03, 06, 07, 08, 12, or 14).
2. Run the final detector, tracker, and (if used) ReID model on all frames.
3. Choose a target pedestrian whose identity is retained as long as possible, preferably through at least one occlusion.
4. Render a video that highlights the target and de-emphasizes all other pedestrians.
5. Report processing speed (frames per second) and the hardware used.

**Note.** MOT16 test annotations are not public, so this notebook provides qualitative evidence only. Quantitative results come from the labeled training sequences in Notebooks 04 and 05.
