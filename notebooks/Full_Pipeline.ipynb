### Full Pipeline: End-to-End Multi-Pedestrian Tracking on MOT16

**Objective.** This notebook presents the complete pipeline in order: dataset preparation, pedestrian detection, appearance-based re-identification, multi-object tracking, and evaluation. It reproduces the project's main results from start to finish.

**Contents**
1. Setup and configuration
2. Data preparation (MOT16 to YOLO format, sequence-level split)
3. Detection (fine-tuning and baseline comparison)
4. Re-identification
5. Tracking and quantitative evaluation (HOTA, IDF1, MOTA)
6. Test-sequence demonstration

**Reproducibility.** Training steps are controlled by flags (for example `RUN_TRAINING`). When a flag is off, the notebook loads saved weights from Drive instead, so the pipeline can be executed without re-training. Model weights are not included in the code submission because of their size.
